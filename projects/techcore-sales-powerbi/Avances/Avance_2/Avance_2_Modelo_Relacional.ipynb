# 🧩 Avance 2: Diseño e Implementación del **Modelo Relacional**

## 📘 Consigna
En esta segunda fase del proyecto para **TechCore**, se diseña e implementa un **modelo relacional** a partir del dataset limpio (`ventasTransformed.csv`).  
El propósito es estructurar las tablas de manera que representen adecuadamente las relaciones entre las entidades del proceso de ventas, manteniendo la **integridad referencial**, la **normalización** y la **coherencia lógica** del modelo.

A diferencia de un esquema Copo de Nieve, el **modelo relacional** busca representar las entidades principales con sus claves primarias y foráneas, sin una descomposición excesiva de las dimensiones.

---

## 🏗️ Tablas a Construir

### **Tablas Dimensionales**
- 🏙️ **Ciudades**  
- 🏢 **Sucursales**  
- 👨‍💼 **Vendedores**  
- 👥 **Clientes**  
- 📦 **Productos**  
- 💳 **Métodos de Pago**

### **Tablas de Hechos**
- 🧾 **Facturas**  
- 📄 **DetalleFacturas**

### **Tablas Derivadas o de Análisis**
- ⭐ **Marcas más vendidas**  
- 🔝 **Productos más vendidos**

---

## 📊 Resultado Esperado
El resultado final será el archivo **`modeloVentas.xlsx`**, que contendrá todas las tablas del **modelo relacional** listas para su importación y modelado en **Power BI**, permitiendo analizar el desempeño de ventas, productos, clientes y sucursales de **TechCore**.


## 1.Carga de Datos

```python
# CODE CELL 1: Importación de Librerías y Carga de Datos

import pandas as pd
import numpy as np
import io


# ==============================================================================
# 0. CONFIGURACIÓN E IMPORTACIÓN DE DATOS
# ==============================================================================

# Cargar el dataset limpio exportado desde Power Query.
# Utilizamos el delimitador ';' (punto y coma).
df_ventas = pd.read_csv(r'C:\Users\DODY DUEÑAS\Documents\Modulo 3\ProyectoVentas_Dody_Salim_Dueñas_Remache\Avance_1\ventasTransformed.csv', sep=';', encoding='latin-1')
print("Dataset 'ventasTransformed.csv' cargado exitosamente.")
```

## 2. Generación de Tablas Dimensionales (DIM)

Se aplica la normalización extrayendo las entidades únicas del negocio y asignando una Clave Primaria (PK) autoincremental a cada dimensión.

###  Tabla de Ciudades

```python
# CODE CELL 2: 1.1 Ciudades (DIM) - PK: CiudadID

# ------------------------------------------------------------------------------
# 1.1 Ciudades (Dim_Ciudades)
# ------------------------------------------------------------------------------
df_ciudades = df_ventas[['CiudadSucursal']].drop_duplicates().reset_index(drop=True)
df_ciudades = df_ciudades.rename(columns={'CiudadSucursal': 'NombreCiudad'})
# Generar la Clave Primaria (PK).
df_ciudades.insert(0, 'CiudadID', df_ciudades.index + 1)
df_ciudades.set_index('CiudadID', inplace=True)
print("Tabla Ciudades creada.")
df_ciudades
```

## Tabla de Vendedores

```python
# CODE CELL 3: 1.2 Vendedores (DIM) - PK: VendedorID

# ------------------------------------------------------------------------------
# 1.2 Vendedores (Dim_Vendedores)
# ------------------------------------------------------------------------------
df_vendedores = df_ventas[['VendedorNombre']].drop_duplicates().reset_index(drop=True)
# Generar la Clave Primaria (PK).
df_vendedores.insert(0, 'VendedorID', df_vendedores.index + 1)
df_vendedores.set_index('VendedorID', inplace=True)
print("Tabla Vendedores creada.")
df_vendedores
```

## Tabla de Clientes

```python
# CODE CELL 4: 1.3 Clientes (DIM) - PK: ClienteID

# ------------------------------------------------------------------------------
# 1.3 Clientes (Dim_Clientes)
# ------------------------------------------------------------------------------
# Columnas que definen un cliente único.
cliente_cols = ['Cliente_Nombre', 'GeneroCliente', 'edadcliente', 'EmailCliente', 'TelefonoCliente', 'Direccion_Cliente']
df_clientes = df_ventas[cliente_cols].drop_duplicates().reset_index(drop=True)
# Generar la Clave Primaria (PK).
df_clientes.insert(0, 'ClienteID', df_clientes.index + 1)
df_clientes.set_index('ClienteID', inplace=True)
print("Tabla Clientes creada.")
df_clientes
```

## Tabla de Sucursales

```python
# CODE CELL 5: 1.4 Sucursales (DIM) - PK: SucursalID, FK: CiudadID

# ------------------------------------------------------------------------------
# 1.4 Sucursales (Dim_Sucursales) - Relación Copo de Nieve (Snowflake)
# ------------------------------------------------------------------------------
df_sucursales = df_ventas[['SucursalNombre', 'CiudadSucursal']].drop_duplicates().reset_index(drop=True)
df_sucursales.insert(0, 'SucursalID', df_sucursales.index + 1)

# Añadir la Clave Foránea (FK) 'CiudadID' mediante JOIN.
df_sucursales = pd.merge(df_sucursales, df_ciudades.reset_index(),
                         left_on='CiudadSucursal', right_on='NombreCiudad', how='left')
# Seleccionar las columnas finales: PK, Nombre, FK.
df_sucursales = df_sucursales[['SucursalID', 'SucursalNombre', 'CiudadID']]
df_sucursales.set_index('SucursalID', inplace=True)
print("Tabla Sucursales creada (normalizada con FK a Ciudades).")
df_sucursales
```

## Tabla de Productos 

```python
# CODE CELL 6: 1.5 Productos (DIM) - PK: ProductoID

# ------------------------------------------------------------------------------
# 1.5 Productos (Dim_Productos)
# ------------------------------------------------------------------------------
# Se desnormalizan y apilan las 3 columnas de producto en una lista maestra.
productos_data = []
for i in range(1, 4):
    cols = [f'Producto{i}Nombre', f'Producto{i}Marca', f'Producto{i}PrecioUnitario']
    temp_df = df_ventas[cols].copy()
    temp_df.columns = ['NombreProducto', 'MarcaProducto', 'PrecioUnitario']
    productos_data.append(temp_df.dropna(subset=['NombreProducto']))

df_productos_raw = pd.concat(productos_data).drop_duplicates()
# Consolidar productos por Nombre y Marca.
df_productos = df_productos_raw.groupby(['NombreProducto', 'MarcaProducto']).agg({
    'PrecioUnitario': 'first'
}).reset_index()

# Generar la Clave Primaria (PK).
df_productos.insert(0, 'ProductoID', df_productos.index + 1)
df_productos.set_index('ProductoID', inplace=True)
print("Tabla Productos creada.")
df_productos
```

## Tabla de Métodos de Pago

```python
# ------------------------------------------------------------------------------
# 1.6 Métodos de Pago (Dim_MetodosPago)
# ------------------------------------------------------------------------------
# Extrae los valores únicos de la columna 'MetodoPago' para crear la tabla maestra.
metodos_pago_unicos = df_ventas['MetodoPago'].unique()

# Crea el DataFrame de la Dimensión
df_metodos_pago = pd.DataFrame({
    'MetodoPagoID': range(1, len(metodos_pago_unicos) + 1),
    'TipoMetodoPago': metodos_pago_unicos
})
df_metodos_pago.set_index('MetodoPagoID', inplace=True) # Mantenemos el PK como índice en Pandas

print("Tabla 1: Métodos de Pago (Dimensión)")
# Corrección: Dejamos la variable sola para que Jupyter use su formato de impresión con el índice PK
df_metodos_pago
```

## 3. Generación de Tablas de Hechos (Fact)

Se crean las tablas transaccionales Facturas (cabecera) y DetalleFacturas (líneas), mapeando las dimensiones a través de las Claves Foráneas (FKs).

## Tabla de Facturas 

```python
# ------------------------------------------------------------------------------
# 2.1 Facturas (Fact_Facturas) - PK: FacturaID, FKs: MetodoPagoID sin _FK
# ------------------------------------------------------------------------------
# Inicializar la tabla de hechos con la cabecera de la venta.
df_facturas = df_ventas.copy().rename(columns={'VentaID': 'FacturaID'})

# Estandarizar nombres de columnas para evitar errores de codificación (ñ, tildes, etc.)
# IMPORTANTE: Eliminada la conversión de 'AñoVenta' a 'AnioVenta'.
import unicodedata
df_facturas.columns = [unicodedata.normalize('NFKC', c).replace('ń', 'ñ') for c in df_facturas.columns]

# Mapear Claves Foráneas (FKs) desde las dimensiones:

# 1. Vendedor
df_facturas = pd.merge(df_facturas, df_vendedores.reset_index(), on='VendedorNombre', how='left')

# 2. Cliente
cliente_map_cols = [
    'Cliente_Nombre', 'GeneroCliente', 'edadcliente',
    'EmailCliente', 'TelefonoCliente', 'Direccion_Cliente'
]
df_facturas = pd.merge(
    df_facturas,
    df_clientes.reset_index()[['ClienteID'] + cliente_map_cols],
    on=cliente_map_cols,
    how='left'
)

# 3. Sucursal
df_facturas = pd.merge(
    df_facturas,
    df_sucursales.reset_index()[['SucursalID', 'SucursalNombre']],
    on='SucursalNombre',
    how='left'
)

# 4. MetodoPago (Mapear y añadir MetodoPagoID)
df_facturas = pd.merge(
    df_facturas,
    df_metodos_pago.reset_index(), # Usa la tabla de dimensión
    left_on='MetodoPago',
    right_on='TipoMetodoPago',
    how='left'
)


# --- RENOMBRADO FINAL DE FKs SIN SUFIJO '_FK' ---
df_facturas.rename(columns={
    'VendedorID': 'VendedorID',     # VendedorID (PK) se convierte en VendedorID (FK)
    'ClienteID': 'ClienteID',       # ClienteID (PK) se convierte en ClienteID (FK)
    'SucursalID': 'SucursalID',     # SucursalID (PK) se convierte en SucursalID (FK)
    'MetodoPagoID': 'MetodoPagoID' # MetodoPagoID (PK) se convierte en MetodoPagoID (FK)
}, inplace=True)


# Seleccionar las columnas finales de la cabecera.
# El nombre 'AñoVenta' se mantiene tal como está en el DataFrame original.
facturas_cols_final = [
    'FacturaID', 'FechaVenta', 'HoraVenta',
    'SucursalID', 'ClienteID', 'VendedorID',
    'MetodoPagoID', 
    'DescuentoVenta', 'TotalVenta', 'MesVenta', 'MesNumero', 'AñoVenta' # <-- 'AñoVenta' REESTABLECIDO
]

# Selección final de columnas y configuración de índice
# Seleccionamos solo las columnas que existen después de todos los merges.
df_facturas = df_facturas[[c for c in facturas_cols_final if c in df_facturas.columns]]
df_facturas.set_index('FacturaID', inplace=True)

print("Tabla Facturas (Cabecera) creada correctamente con 'AñoVenta'.")
df_facturas
```

## Tabla de Detalles de Facturas

```python
# CODE CELL 8: 2.2 DetalleFacturas (FACT Detail)

# ------------------------------------------------------------------------------
# 2.2 DetalleFacturas (Fact_DetalleFacturas) - PK: DetalleID, FKs: FacturaID, ProductoID
# ------------------------------------------------------------------------------
detalle_data = []
# Desnormalización de las líneas de detalle (se crea una fila por cada producto vendido).
for i in range(1, 4):
    cols = [
        'VentaID', f'Producto{i}Nombre', f'Producto{i}Marca',
        f'Producto{i}Cantidad', f'Producto{i}PrecioUnitario', f'Producto{i}Subtotal'
    ]
    temp_df = df_ventas.dropna(subset=[f'Producto{i}Nombre'])[cols].copy()
    temp_df.columns = ['FacturaID', 'NombreProducto', 'MarcaProducto', 'Cantidad', 'PrecioUnitario', 'Subtotal']
    detalle_data.append(temp_df)

df_detalle_facturas = pd.concat(detalle_data, ignore_index=True)

# Mapeo de ProductoID (FK) uniendo con la dimensión Productos.
df_detalle_facturas = pd.merge(df_detalle_facturas, df_productos.reset_index()[['ProductoID', 'NombreProducto', 'MarcaProducto']],
                               on=['NombreProducto', 'MarcaProducto'], how='left')

# Generar DetalleID (PK) para cada línea transaccional.
df_detalle_facturas.insert(0, 'DetalleID', df_detalle_facturas.index + 1)
# Seleccionar las columnas finales (PK, FKs y métricas).
detalle_facturas_cols = [
    'DetalleID', 'FacturaID', 'ProductoID', 'Cantidad', 'PrecioUnitario', 'Subtotal'
]
df_detalle_facturas = df_detalle_facturas[detalle_facturas_cols]
df_detalle_facturas.set_index('DetalleID', inplace=True)
print("Tabla DetalleFacturas (Línea) creada.")
df_detalle_facturas
```

## 4. Verificación de Consistencia y Control de Calidad

Se generan los reportes exploratorios solicitados para validar que el modelo relacional funciona correctamente y los datos son consistentes.

```python
# CODE CELL 9: Generación de Reportes

# ------------------------------------------------------------------------------
# 3.1 Reporte: Total de ventas por marca.
# ------------------------------------------------------------------------------
df_ventas_por_marca = pd.merge(df_detalle_facturas.reset_index(), df_productos.reset_index()[['ProductoID', 'MarcaProducto']], on='ProductoID', how='left')
df_ventas_por_marca = df_ventas_por_marca.groupby('MarcaProducto')['Subtotal'].sum().sort_values(ascending=False).to_frame(name='TotalVentas').reset_index()

# ------------------------------------------------------------------------------
# 3.2 Reporte: Top 10 productos más vendidos (por cantidad).
# ------------------------------------------------------------------------------
df_productos_vendidos = pd.merge(df_detalle_facturas.reset_index(), df_productos.reset_index()[['ProductoID', 'NombreProducto', 'MarcaProducto']], on='ProductoID', how='left')
df_top_productos = df_productos_vendidos.groupby(['NombreProducto', 'MarcaProducto'])['Cantidad'].sum().sort_values(ascending=False).head(10).to_frame(name='CantidadTotalVendida').reset_index()

print("\n--- Resultados de los Reportes de Consistencia ---")
print("\nTotal de ventas por marca:")
print(df_ventas_por_marca.to_markdown(index=False))
print("\nTop 10 productos más vendidos (por cantidad):")
print(df_top_productos.to_markdown(index=False))
```

## Tablas Derivadas o de Análisis

## Productos más vendidos

```python
# -------------------------------------------------------------------
# --- 2. Preparación de Datos (Reestructuración/Unpivot) ---
# Necesitamos una sola columna de 'ProductoNombre', 'Marca', y 'Cantidad'
# para poder hacer las sumas de ventas correctamente.

# 1. Extraer y renombrar los datos para cada Producto (Producto 1, 2 y 3)
productos1 = df_ventas[['VentaID', 'Producto1Nombre', 'Producto1Marca', 'Producto1Cantidad']].copy()
productos1.columns = ['VentaID', 'ProductoNombre', 'Marca', 'Cantidad']

productos2 = df_ventas[['VentaID', 'Producto2Nombre', 'Producto2Marca', 'Producto2Cantidad']].copy()
productos2.columns = ['VentaID', 'ProductoNombre', 'Marca', 'Cantidad']

productos3 = df_ventas[['VentaID', 'Producto3Nombre', 'Producto3Marca', 'Producto3Cantidad']].copy()
productos3.columns = ['VentaID', 'ProductoNombre', 'Marca', 'Cantidad']

# 2. Unir (Concatenar) los tres DataFrames en uno solo
df_productos_vendidos = pd.concat([productos1, productos2, productos3], ignore_index=True)

# 3. Limpiar: Filtrar filas donde la Cantidad es 0 o donde el ProductoNombre es nulo
df_productos_vendidos = df_productos_vendidos[df_productos_vendidos['Cantidad'] > 0].dropna(subset=['ProductoNombre', 'Marca'])
```

```python
# -------------------------------------------------------------------
# --- 3. Creación de la Vista: Productos más vendidos (Análisis) ---

# 1. Agrupar por el nombre del producto y sumar la cantidad vendida.
df_prod_vendidos_view = df_productos_vendidos.groupby('ProductoNombre')['Cantidad'].sum().reset_index()
df_prod_vendidos_view.rename(columns={'Cantidad': 'TotalUnidadesVendidas'}, inplace=True)

# 2. Ordenar por la cantidad vendida y generar el ranking.
df_prod_vendidos_view = df_prod_vendidos_view.sort_values(by='TotalUnidadesVendidas', ascending=False)
df_prod_vendidos_view['Ranking'] = df_prod_vendidos_view['TotalUnidadesVendidas'].rank(method='min', ascending=False).astype(int)

# 3. Seleccionar y reordenar las columnas finales.
df_prod_vendidos_view = df_prod_vendidos_view[['Ranking', 'ProductoNombre', 'TotalUnidadesVendidas']]

print("\n Tabla: Productos más vendidos")
df_prod_vendidos_view
```

## Marcas mas Vendidas

```python
# -------------------------------------------------------------------
# --- 4. Creación de la Vista: Marcas más vendidas (Análisis) ---

# 1. Agrupar la tabla de ventas "plana" por la Marca y sumar la cantidad vendida.
df_marcas_vendidas_view = df_productos_vendidos.groupby('Marca')['Cantidad'].sum().reset_index()
df_marcas_vendidas_view.rename(columns={'Cantidad': 'TotalUnidadesVendidas'}, inplace=True)

# 2. Ordenar por la cantidad vendida y generar el ranking.
df_marcas_vendidas_view = df_marcas_vendidas_view.sort_values(by='TotalUnidadesVendidas', ascending=False)
df_marcas_vendidas_view['Ranking'] = df_marcas_vendidas_view['TotalUnidadesVendidas'].rank(method='min', ascending=False).astype(int)

# 3. Seleccionar y reordenar las columnas finales.
df_marcas_vendidas_view = df_marcas_vendidas_view[['Ranking', 'Marca', 'TotalUnidadesVendidas']]

print("Tabla: Marcas más vendidas (Análisis/Vista)")
df_marcas_vendidas_view
```

## Exportación a Excel

```python
# ------------------------------------------------------------------------------
# 4. EXPORTACIÓN DEL MODELO FINAL (Corregida a 8 Tablas Base para Power BI)
# ------------------------------------------------------------------------------

# Se agrupan las 8 Tablas Base del Modelo Relacional (Hechos y Dimensiones)
# NOTA: df_marcas se ha EXCLUIDO de este diccionario, como solicitaste.
dataframes_base_model = {
    'Facturas': df_facturas.reset_index(drop=False), # Exporta FacturaID
    'DetalleFacturas': df_detalle_facturas.reset_index(drop=False), # Exporta DetalleID
    'Productos': df_productos.reset_index(drop=False), # Exporta ProductoID
    'Clientes': df_clientes.reset_index(drop=False), # Exporta ClienteID
    'Vendedores': df_vendedores.reset_index(drop=False), # Exporta VendedorID
    'Sucursales': df_sucursales.reset_index(drop=False), # Exporta SucursalID
    'Ciudades': df_ciudades.reset_index(drop=False), # Exporta CiudadID
    'MetodosPago': df_metodos_pago.reset_index(drop=False) # Exporta MetodoPagoID
}

# Crear el archivo Excel 'modeloVentas.xlsx' con cada DataFrame como una hoja.
with pd.ExcelWriter('modeloVentas.xlsx', engine='xlsxwriter') as writer:
    
    # Exportar las 8 Tablas Base del Modelo Relacional
    for sheet_name, df in dataframes_base_model.items():
        # Exportar sin el índice de pandas (pero incluyendo la columna PK).
        df.to_excel(writer, sheet_name=sheet_name, index=False)
        
    # Exportar las 2 Tablas de Análisis/Vistas en hojas separadas
    # Estas vistas utilizan df_marcas, pero se exportan separadamente.
    df_prod_vendidos_view.to_excel(writer, sheet_name='ProductosMasVendidos', index=False)
    df_marcas_vendidas_view.to_excel(writer, sheet_name='MarcasMasVendidas', index=False)
    

print("\n------------------------------------------------------------------")
print("PROCESO DE MODELADO Y EXPORTACIÓN FINALIZADO.")
print("Archivo 'modeloVentas.xlsx' generado con las 8 tablas base del Modelo Relacional y 2 Vistas de Análisis.")
print("Las claves primarias/foráneas están correctamente formateadas para Power BI.")
print("------------------------------------------------------------------")
```
